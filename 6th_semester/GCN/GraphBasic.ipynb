## GRAPH LABELING TECHNIQUES

```python
import matplotlib 

matplotlib.use('TKAgg') # Use the TkAgg backend for rendering

import matplotlib.pyplot as plt # use to draw graphs on screen 
import numpy as np
import networkx as nx  # use to create, manage and visualize  graphs


from matplotlib.patches  import Rectangle
import matplotlib.patches as mpatches


print("=" * 70)
print("20 EXAMPLES: GRAPHS  AND NETWORKS")
print("=" * 70)
```

#### Exercise 1 — Simple Node Labels

```python
G1 = nx.cycle_graph(8)  # Create a cycle graph with 8 nodes

plt.figure(figsize = (8,5)) # Set the figure size for the plot

nx.draw(
    G1,
    with_labels = True,         #show labels
    node_color = 'aqua',        # set node color
    node_size = 500,
    font_size = 12,             #Label font size
    font_color = 'black',
    font_weight = 'bold',
)

plt.title("Default Node labels(Numbers)")
plt.show(block = True)

```

#### Exercise 2 — Custom Node Labels

```python
import networkx as nx
import matplotlib.pyplot as plt

# step by step : 1. Create graph 2. Define custom labels 3. Apply custom labels

G2 = nx.Graph()  # create empty undirected graph

G2.add_edges_from([
    (1, 2),
    (2, 3),
    (3, 1),
    (4, 3)
])

custom_labels = {
    1: 'A',
    2: 'B',
    3: 'C',
    4: 'D'
}

plt.figure(figsize=(8,6))

pos = nx.spring_layout(G2, seed = 42) #NetworkX decides where to place the nodes.
# seed=42 → gives a consistent layout each time.

# Draw a graph without labels first
nx.draw(G2, pos, node_color = 'red', node_size= 500)

# Add custom labels
nx.draw_networkx_labels(
    G2,
    pos, 
    labels = custom_labels,
    font_size = 16,
    font_weight = 'bold',
    font_color = 'darkgreen'
)

plt.title("Custom Node labels")
plt.show()



```

#### Exercise 3 — Different Label Positions (Labels Above Nodes)

```python
# Step by step : 1. create graph 2. position labels above nodes 3. different alignment

import networkx as nx 
import matplotlib.pyplot as plt 

G3 = nx.star_graph(5)  # Creates a star-shaped graph.

# One central node is connected to the other nodes.

plt.figure(figsize = (8,6))

pos = nx.spring_layout(G3, seed = 42) # NetworkX calculates where each node should appear.

nx.draw_networkx_nodes(
    G3, pos, node_color = 'lightcoral', node_size = 5000
)
nx.draw_networkx_edges(G3, pos)


# The f'N{node}' means: Put N before the node number.

labels = {
    node: f'N{node}' for node in G3.nodes()
}

label_pos = {
    node: (x,y + 0.05) #label moves slightly upward.
    for node, (x,y) in pos.items()
}

nx.draw_networkx_labels(
        G3,
        label_pos,
        labels = labels,
        font_size = 12
    )



plt.title("Example 3: Labels Above Nodes")
plt.show()





```

#### Exercise 4 — Edge Labels (Weights)

```python
import networkx as nx
import matplotlib.pyplot as plt 


# STEP BY STEP: 1. Create graph with weights 2. Extract edge labels 3. Display edge

G4 = nx.Graph()

G4.add_weighted_edges_from([
    (1, 2, 5),
    (2, 3, 3),
    (3, 4, 7),
    (4, 1, 2),
    (1, 3, 4)
])

plt.figure(figsize = (8,6))

pos = nx.spring_layout(G4, seed = 42)

# Draw graph
nx.draw(G4, pos, with_labels=True, node_color='lightgray',node_size=500, font_size=14)

# Add edge labels (weights)

edge_labels = nx.get_edge_attributes(G4, 'weight')

nx.draw_networkx_edge_labels(G4,pos, edge_labels = edge_labels, font_size= 12, font_color='red' )


plt.title("Example 4: Edge Labels (weights)")
plt.show(block = True)

```

#### Exercise 5 - Custom Edge Labels

```python
import networkx as nx 
import matplotlib.pyplot as plt

# Creates an empty undirected graph.
G5 = nx.Graph()

# Add edges
G5.add_edges_from([
    ('A', 'B'),
    ('B', 'C'),
    ('C', 'D'),
    ('D', 'A')
])

custom_edge_label = {
    ('A', 'B'):'Friend',
    ('B', 'C'):'Colleague',
    ('C', 'D'):'Family',
    ('D', 'A'):'Neighbor'
}

plt.figure(figsize=(8,6))

pos = nx.spring_layout(G5, seed=42)

nx.draw(G5, pos,
        with_labels=True,
        node_color = 'lightgreen',
        node_size = 500,
        font_size = 14,
        font_weight = 'bold'
        )

nx.draw_networkx_edge_labels(
    G4,
    pos,
    edge_labels= custom_edge_label,
    font_size = 11,
    font_color='blue'
)

plt.title("Example 5: Custom Edge Labels")
plt.show(block = True)
```

#### Exercise 6 - Node Labels with Attributes

```python
import networkx as  nx 
import matplotlib.pyplot as plt

# STEP BY STEP: 1. Create graph with node attributes 2. Create labels from attributes 3. Display attribute labels

G6 = nx.Graph()

G6.add_node(1, name='Alice', age=25)
G6.add_node(2, name='Bob', age=30)
G6.add_node(3, name='Charlie', age=35)
G6.add_node(4, name='Diana', age=28)

G6.add_edges_from([
    (1, 2),
    (2, 3),
    (3, 4),
    (4, 1)
])

node_labels = {
    node: f"{G6.nodes[node]['name']}\n({G6.nodes[node]['age']})"
    for node in G6.nodes()
}

plt.figure(figsize=(8,6))

pos = nx.spring_layout(G6, seed = 42)

nx.draw(
        G6,
        pos, 
        node_color= 'lightcoral',
        node_size=(1500)
        )

nx.draw_networkx_labels(
    G6,
    pos,
    labels = node_labels,
    font_size = 10,
    
)

plt.title("Example 6: Node Lables with Attrivures")
plt.show(block=True)
```

#### Exercise 7 — Combined Node and Edge Labels

```python
import networkx as nx
import matplotlib.pyplot as plt

G7 = nx.Graph()

G7.add_nodes_from(['A', 'B', 'C', 'D'])

G7.add_weighted_edges_from([
    ('A', 'B', 5),
    ('B', 'C', 3),
    ('C', 'D', 7),
    ('D', 'A', 2)
])

plt.figure(figsize=(8, 6))

pos = nx.spring_layout(G7, seed=42)

nx.draw(
    G7, pos,
    node_color='lightyellow',
    node_size=700
)

node_labels = {
    node: f"★ {node}"
    for node in G7.nodes()
}

nx.draw_networkx_labels(
    G7, pos,
    labels=node_labels,
    font_size=14
)

edge_labels = nx.get_edge_attributes(G7, 'weight')
edge_labels = {
    edge: f"w={weight}"
    for edge, weight in edge_labels.items()
}

nx.draw_networkx_edge_labels(
    G7, pos,
    edge_labels=edge_labels,
    font_size=10,
    font_color='blue'
)

plt.title("Example 7: Combined Node and Edge Labels")
plt.show()
```
