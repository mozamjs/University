#### Exercise 1: Create a list of integers from 1 to 5 and apply some operation also print.

```python
arr = [1,2,3,4,5]

print(arr[2])

arr[2]= 99
print(arr)

arr.append(60)
print(arr)

arr.remove(20)
print(arr)

print(len(arr))

for i in arr:
    print(i,end=" ")

print("\n")

for i in range(len(arr)):
    print(arr[i],end=" ")
```

#### Exercise 2: Traversing array 


```python
arr = [12,7,19,4,25]
# by value
for x in arr:
    print(x)
# by index
for i in range(len(arr)):
    print(i, arr[i], end=" ")

print("\n")

# index + value both using enumerate() 
print ("Index  value")
for i,x in enumerate(arr):
    print(i, x )
```

#### Exercise 3: array.array


```python
import array 

val = array.array('i',[1,2,3,4,5,6,7,8])

for i in range(0,6):
    print(val[i], end=" ")

print("\n")

for i in range (0, len(val)):
    print(val[i], end =" ")

print("\n")

for x in val: 
    print(x, end=",")

print("\n")

val.reverse()

print(val)

for i in range(0,len(val)):
    print(val[i], end = " ")
```

#### Exercise 4: Insert Element

```python
# from array import array
from array import *

# i for integer, f for float, d for double, u for unicode character

arr = array('i',[10,20,30,40,50])
# arr = array('f', [1.2, 2.5, 3.7])
print(arr)

# insert element at a specific position
arr.insert(2,99)
print(arr)

# insert element at the end of the array
arr.append(987)
print(arr)

# Update an element at a specific index
arr[3] = 70
print(arr)

copyArray = array(arr.typecode, (x for x in arr))

for i in range(0, len(arr)):
    print(arr[i], end = " ")

print("\n")

for i in range(0 , len(copyArray)):
    print(copyArray[i], end = " ")
```

#### Copy array

```python
from array import array

arr = array('i',[1,2,3,4,5,6,7,8])

copyArray = array(arr.typecode, (x*2 for x in arr) )

print("orignal array:", arr)
print("Copy array with elements multiplied by 2: ", copyArray)

copyArray.pop(3)
print("After removing element at index 3: ", copyArray)
copyArray.remove(12)
print("After removing element 12: ", copyArray)

```

#### Exercise 6: Slicing

```python
from array import array 

arr = array('i', [1,2,3,4,5,6,7,8])

# [start : stop : step]

a = arr[0:3]
print(a)
b = arr[1::2]
print(b)
c = arr[:6]
print(c)
d = arr[::-1]
print(d)

for i in range(0, len(arr)):
    print(arr[i], end = " ")
```

#### Exercise 7: Taking input form user
